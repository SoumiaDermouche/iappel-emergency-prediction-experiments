# Operator D — Sensitivity Analysis (with vs. without Operator D)

This notebook reports the operator-wise evaluation scores and prediction accuracy, computed **with** and **without** Operator D, to assess the extent to which this apparent outlier skews the results (Reviewer #3 request).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

file_path = "evaluations.csv"
df = pd.read_csv(file_path)
df.columns = df.columns.str.strip()

rename_map = {
    'Transcription': 'Test',
    'Fidélité': 'Transcription',
    'Note résumé': 'Summary',
    'Note entité': 'NER',
    'Urgence correcte': 'Emergency',
    'Catégorie correcte': 'Emergency category'
}
df.rename(columns=rename_map, inplace=True)

for col in ['Transcription', 'Summary', 'NER']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['Emergency_bin'] = df['Emergency'].map(
    lambda x: 1 if str(x).strip().lower() == 'oui' else (0 if str(x).strip().lower() == 'non' else np.nan)
)
df['Category_bin'] = df['Emergency category'].map(
    lambda x: 1 if str(x).strip().lower() == 'oui' else (0 if str(x).strip().lower() == 'non' else np.nan)
)

# WITH Operator D: analysis performed over the ENTIRE file (all 157 calls),
# regardless of whether the operator identity is known for a given call.
op_df = df.copy()

# WITHOUT Operator D: analysis performed over the entire file MINUS the 7 calls
# annotated by Operator D (157 - 7 = 150 calls). Calls with an unknown operator
# identity are kept (they are not Operator D's calls); only the 7 calls
# explicitly labeled 'D' are removed.
op_df_noD = df[df['Operateur'] != 'D'].copy()

n_total = len(df)
n_D = len(df[df['Operateur'] == 'D'])
print(f"WITH Operator D    : {len(op_df)} calls (full file)")
print(f"WITHOUT Operator D : {len(op_df_noD)} calls ({n_total} - {n_D} = {len(op_df_noD)})")
```

## 0. Correlation heatmap and score distributions — With vs. Without Operator D

All plots in this section are shown twice, side by side: computed over the full 157-call file (**With Operator D**) and over the 150 calls obtained after removing Operator D's 7 annotated calls (**Without Operator D**).

```python
import os
os.makedirs("figures", exist_ok=True)

score_cols = ['Transcription', 'Summary', 'NER']
corr_withD = df[score_cols].corr()

fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(corr_withD.values, cmap="coolwarm", vmin=0.80, vmax=1.0)
ax.set_xticks(np.arange(len(corr_withD.columns)))
ax.set_yticks(np.arange(len(corr_withD.index)))
ax.set_xticklabels(corr_withD.columns, rotation=45, ha="right")
ax.set_yticklabels(corr_withD.index)
for i in range(len(corr_withD.index)):
    for j in range(len(corr_withD.columns)):
        ax.text(j, i, f"{corr_withD.iloc[i, j]:.2f}", ha="center", va="center", color="black")
plt.colorbar(im)
ax.set_title("Correlation Heatmap — With Operator D (n=157)")
plt.tight_layout()
plt.savefig("figures/correlation_heatmap_with_D.png", dpi=300, bbox_inches="tight")
plt.show()

print("Correlation matrix — With Operator D (n=157):")
print(corr_withD)

d1 = df[['Transcription', 'Summary']].dropna()
r_summary, p_summary = stats.pearsonr(d1['Transcription'], d1['Summary'])
d2 = df[['Transcription', 'NER']].dropna()
r_ner, p_ner = stats.pearsonr(d2['Transcription'], d2['NER'])
print(f"Pearson (Transcription, Summary): r = {r_summary:.3f}, p = {p_summary:.2e}, n = {len(d1)}")
print(f"Pearson (Transcription, NER)    : r = {r_ner:.3f}, p = {p_ner:.2e}, n = {len(d2)}")
```

```python
os.makedirs("figures", exist_ok=True)

corr_noD = op_df_noD[score_cols].corr()

fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(corr_noD.values, cmap="coolwarm", vmin=0.80, vmax=1.0)
ax.set_xticks(np.arange(len(corr_noD.columns)))
ax.set_yticks(np.arange(len(corr_noD.index)))
ax.set_xticklabels(corr_noD.columns, rotation=45, ha="right")
ax.set_yticklabels(corr_noD.index)
for i in range(len(corr_noD.index)):
    for j in range(len(corr_noD.columns)):
        ax.text(j, i, f"{corr_noD.iloc[i, j]:.2f}", ha="center", va="center", color="black")
plt.colorbar(im)
ax.set_title("Correlation Heatmap — Without Operator D (n=150)")
plt.tight_layout()
plt.savefig("figures/correlation_heatmap_without_D.png", dpi=300, bbox_inches="tight")
plt.show()

print("Correlation matrix — Without Operator D (n=150):")
print(corr_noD)

d1 = op_df_noD[['Transcription', 'Summary']].dropna()
r_summary, p_summary = stats.pearsonr(d1['Transcription'], d1['Summary'])
d2 = op_df_noD[['Transcription', 'NER']].dropna()
r_ner, p_ner = stats.pearsonr(d2['Transcription'], d2['NER'])
print(f"Pearson (Transcription, Summary): r = {r_summary:.3f}, p = {p_summary:.2e}, n = {len(d1)}")
print(f"Pearson (Transcription, NER)    : r = {r_ner:.3f}, p = {p_ner:.2e}, n = {len(d2)}")
```

```python
os.makedirs("figures", exist_ok=True)
df['Emergency_cat'] = df['Emergency_bin'].map({0: 'No', 1: 'Yes'})

for col, ylabel in [('Transcription', 'Transcription Score (1 to 5)'),
                     ('NER', 'NER Score (1 to 5)'),
                     ('Summary', 'Summary Score (1 to 5)')]:
    plt.figure(figsize=(6, 5))
    sns.boxplot(x='Emergency_cat', y=col, data=df, palette='Set2')
    plt.title(f"{col} Score by Emergency Status — With Operator D (n=157)")
    plt.xlabel("Correct urgency prediction")
    plt.ylabel(ylabel)
    plt.ylim(0.5, 5.5)
    plt.tight_layout()
    plt.savefig(f"figures/boxplot_{col.lower()}_with_D.png", dpi=300, bbox_inches="tight")
    plt.show()
```

```python
os.makedirs("figures", exist_ok=True)
op_df_noD['Emergency_cat'] = op_df_noD['Emergency_bin'].map({0: 'No', 1: 'Yes'})

for col, ylabel in [('Transcription', 'Transcription Score (1 to 5)'),
                     ('NER', 'NER Score (1 to 5)'),
                     ('Summary', 'Summary Score (1 to 5)')]:
    plt.figure(figsize=(6, 5))
    sns.boxplot(x='Emergency_cat', y=col, data=op_df_noD, palette='Set2')
    plt.title(f"{col} Score by Emergency Status — Without Operator D (n=150)")
    plt.xlabel("Correct urgency prediction")
    plt.ylabel(ylabel)
    plt.ylim(0.5, 5.5)
    plt.tight_layout()
    plt.savefig(f"figures/boxplot_{col.lower()}_without_D.png", dpi=300, bbox_inches="tight")
    plt.show()
```

## 1. Operator-wise table — WITH Operator D (A, C, D)

*This per-operator breakdown necessarily relies only on the 15 calls with a known operator identity (A: n=5, C: n=3, D: n=7), since operator identity is not available for the rest of the file. The aggregate "WITH Operator D" statistics reported below (overall means, correctness rates) are computed over the full 157-call file.*

```python
rows = []
for o in ['A', 'C', 'D']:
    sub = op_df[op_df['Operateur'] == o]
    rows.append([o, len(sub), sub['Transcription'].mean(), sub['NER'].mean(), sub['Summary'].mean(),
                 sub['Emergency_bin'].mean() * 100, sub['Category_bin'].mean() * 100])

table_with_D = pd.DataFrame(rows, columns=['Operator', 'Nb', 'Trans.', 'NER', 'Summary', '% Emer.', '% Cat.']).round(2)
table_with_D
```

```python
# Operator-level comparison necessarily uses only the calls with known operator
# identity (A: n=5, C: n=3, D: n=7); this cannot be extended to the full file
# since the operator label is unknown for the remaining calls.

print("Kruskal-Wallis tests across operators A, C, D (WITH Operator D):\n" + "-" * 55)

for col in ['Transcription', 'NER', 'Summary']:
    groups = [
        op_df[op_df['Operateur'] == o][col].dropna()
        for o in ['A', 'C', 'D']
    ]
    
    h, p = stats.kruskal(*groups)
    
    print(
        f"{col:15s}: H = {h:.3f}, p = {p:.4f}  "
        f"(n = {[len(g) for g in groups]})"
    )


# ============================================================
# Overall results across all calls
# ============================================================

print(f"\nOverall means across all {len(op_df)} calls (WITH Operator D, full file):")

print(op_df[['Transcription', 'Summary', 'NER']].mean().round(3))


# ============================================================
# Urgency correctness
# Percentage calculated only on non-missing responses
# ============================================================

urgency_valid = op_df['Emergency_bin'].dropna()

urgency_correct = int(urgency_valid.sum())
urgency_total = len(urgency_valid)
urgency_pct = urgency_correct / urgency_total * 100

print(
    f"Urgency correctness  : {urgency_pct:.2f}% "
    f"({urgency_correct}/{urgency_total} valid responses)"
)


# ============================================================
# Category correctness
# Percentage calculated only on non-missing responses
# ============================================================

category_valid = op_df['Category_bin'].dropna()

category_correct = int(category_valid.sum())
category_total = len(category_valid)
category_pct = category_correct / category_total * 100

print(
    f"Category correctness : {category_pct:.2f}% "
    f"({category_correct}/{category_total} valid responses)"
)
```

## 2. Operator-wise table — WITHOUT Operator D (A, C only)

*This per-operator breakdown is restricted to the known-operator calls excluding D (A: n=5, C: n=3). The aggregate "WITHOUT Operator D" statistics reported below (overall means, correctness rates) are computed over 157 − 7 = 150 calls, i.e. the full file minus Operator D's 7 annotated calls.*

```python
rows2 = []
for o in ['A', 'C']:
    sub = op_df_noD[op_df_noD['Operateur'] == o]
    rows2.append([o, len(sub), sub['Transcription'].mean(), sub['NER'].mean(), sub['Summary'].mean(),
                  sub['Emergency_bin'].mean() * 100, sub['Category_bin'].mean() * 100])

table_without_D = pd.DataFrame(rows2, columns=['Operator', 'Nb', 'Trans.', 'NER', 'Summary', '% Emer.', '% Cat.']).round(2)
table_without_D
```

```python
# Only 2 groups remain (A, C) once D is excluded -> use Mann-Whitney U instead of Kruskal-Wallis
# (operator-level comparison still relies on the calls with known operator identity)

print("Mann-Whitney U tests, Operator A vs. Operator C (WITHOUT Operator D):\n" + "-" * 65)

for col in ['Transcription', 'NER', 'Summary']:
    gA = op_df_noD[op_df_noD['Operateur'] == 'A'][col].dropna()
    gC = op_df_noD[op_df_noD['Operateur'] == 'C'][col].dropna()
    
    u, p = stats.mannwhitneyu(gA, gC, alternative='two-sided')
    
    print(
        f"{col:15s}: U = {u:.3f}, p = {p:.4f}  "
        f"(nA={len(gA)}, nC={len(gC)})"
    )


print(f"\nOverall means across the {len(op_df_noD)} remaining calls (WITHOUT Operator D, 157 - 7):")
print(op_df_noD[['Transcription', 'Summary', 'NER']].mean().round(3))


# Urgency correctness
urgency_valid = op_df_noD['Emergency_bin'].dropna()
urgency_correct = int(urgency_valid.sum())
urgency_total = len(urgency_valid)
urgency_pct = urgency_correct / urgency_total * 100

print(
    f"Urgency correctness  : {urgency_pct:.2f}% "
    f"({urgency_correct}/{urgency_total} valid responses)"
)


# Category correctness
category_valid = op_df_noD['Category_bin'].dropna()
category_correct = int(category_valid.sum())
category_total = len(category_valid)
category_pct = category_correct / category_total * 100

print(
    f"Category correctness : {category_pct:.2f}% "
    f"({category_correct}/{category_total} valid responses)"
)
```

## 3. Summary comparison (with vs. without Operator D)

The table below is computed dynamically: **"With Operator D"** uses the full 157-call file, and **"Without Operator D"** uses the 150 calls obtained by removing only Operator D's 7 annotated calls (157 − 7 = 150).

```python
summary = pd.DataFrame({
    'Metric': ['Mean Transcription score', 'Mean Summary score', 'Mean NER score',
               'Urgency correctness (% oui)', 'Category correctness (% oui)', 'N calls'],
    f'With Operator D (n={len(op_df)})': [
        op_df['Transcription'].mean(), op_df['Summary'].mean(), op_df['NER'].mean(),
        op_df['Emergency_bin'].mean() * 100, op_df['Category_bin'].mean() * 100, len(op_df)
    ],
    f'Without Operator D (n={len(op_df_noD)})': [
        op_df_noD['Transcription'].mean(), op_df_noD['Summary'].mean(), op_df_noD['NER'].mean(),
        op_df_noD['Emergency_bin'].mean() * 100, op_df_noD['Category_bin'].mean() * 100, len(op_df_noD)
    ],
}).round(2)
summary
```

**Interpretation:** Comparing the full 157-call file (WITH Operator D) to the 150-call file obtained by removing only Operator D's 7 annotated calls (WITHOUT Operator D) shows how much D's annotations shift the aggregate subjective scores and correctness rates once the operator-labeled calls are re-integrated into the wider dataset. The per-operator Kruskal-Wallis / Mann-Whitney tests above remain restricted to the 15 calls with known operator identity (A: n=5, C: n=3, D: n=7), since operator-level significance testing requires knowing which operator produced each score. Re-run this notebook on `evaluations.csv` to obtain the updated figures.

```python
print("Mann-Whitney U tests — Correct vs. Incorrect urgency prediction (WITH Operator D, n=157):\n" + "-" * 75)
for col in ['Transcription', 'Summary', 'NER']:
    g_correct = df[df['Emergency_cat'] == 'Yes'][col].dropna()
    g_incorrect = df[df['Emergency_cat'] == 'No'][col].dropna()
    u, p = stats.mannwhitneyu(g_correct, g_incorrect, alternative='two-sided')
    print(f"{col:15s}: U = {u:.2f}, p = {p:.2e}  (n_correct={len(g_correct)}, n_incorrect={len(g_incorrect)})")
```

```python
print("Mann-Whitney U tests — Correct vs. Incorrect urgency prediction (WITHOUT Operator D, n=150):\n" + "-" * 75)
for col in ['Transcription', 'Summary', 'NER']:
    g_correct = op_df_noD[op_df_noD['Emergency_cat'] == 'Yes'][col].dropna()
    g_incorrect = op_df_noD[op_df_noD['Emergency_cat'] == 'No'][col].dropna()
    u, p = stats.mannwhitneyu(g_correct, g_incorrect, alternative='two-sided')
    print(f"{col:15s}: U = {u:.2f}, p = {p:.2e}  (n_correct={len(g_correct)}, n_incorrect={len(g_incorrect)})")
```
